```python
import gradio as gr

def sentence_type(sentence):
    sentence = sentence.strip()
    if sentence.endswith('.'):
        return "Розповідне речення"
    elif sentence.endswith('?'):
        return "Питальне речення"
    elif sentence.endswith('!'):
        return "Окличне речення"
    else:
        return "Не вдалося визначити тип речення"

iface1 = gr.Interface(
    fn=sentence_type,
    inputs=gr.Textbox(label="Введіть речення"),
    outputs=gr.Label(label="Тип речення"),
    title="Визначення типу речення",
    description="Програма визначає тип речення за кінцевим розділовим знаком."
)
import re
from collections import Counter

def text_analysis(text):

    words = re.findall(r'\b\w+\b', text.lower())
    sentences = re.split(r'[.!?]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    num_words = len(words)
    num_sentences = len(sentences)
    avg_word_len = round(sum(len(w) for w in words)/num_words, 2) if num_words else 0
    avg_sentence_len = round(sum(len(s.split()) for s in sentences)/num_sentences, 2) if num_sentences else 0


    word_counts = Counter(words)
    top5_words = word_counts.most_common(5)
    unique_words = [w for w, c in word_counts.items() if c == 1]

    result = {
        "Кількість слів": num_words,
        "Кількість речень": num_sentences,
        "Середня довжина слова": avg_word_len,
        "Середня довжина речення (слов)": avg_sentence_len,
        "Топ-5 найчастіших слів": top5_words,
        "Унікальні слова": unique_words
    }
    return result

iface2 = gr.Interface(
    fn=text_analysis,
    inputs=gr.Textbox(label="Введіть текст", lines=5),
    outputs=gr.JSON(label="Результати аналізу"),
    title="Стилістичний аналіз тексту",
    description="Програма аналізує текст, рахує слова, речення, середню довжину та топ-5 слів."
)
demo = gr.TabbedInterface([iface1, iface2], ["Тип речення", "Стилістичний аналіз"])
demo.launch()
```
